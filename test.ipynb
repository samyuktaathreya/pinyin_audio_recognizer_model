```python
import glob, random
import numpy as np
from IPython.display import Audio, display
from data import load_audio, augment_audio, SAMPLE_RATE

files = random.sample(glob.glob("/PATH/TO/processed/*/*.wav"), 5)   # your Mac's processed folder
for f in files:
    wav = load_audio(f)
    print("\n", f.split("/")[-2], f.split("/")[-1], "- original")
    display(Audio(wav, rate=SAMPLE_RATE, normalize=False))
    for i in range(3):
        print("   noisy version", i + 1)
        display(Audio(augment_audio(wav, np.random.default_rng(i)), rate=SAMPLE_RATE, normalize=False))
```
